# Temperature Conversion and Threshhold Checker

```python
# This function converts a temperature from Celcius to Fahrenheit
def celcius_to_fahrenheit(celcius):
    fahrenheit = (celcius * 9/5) + 32
    return fahrenheit

# Ask how many temperatures the user wants to enter
num_temperatures = int(input("How many temperatures do you want to enter? "))

# Create an empty list to store the temperatures
temperatures = []
# Loop through the temperatures
for i in range(num_temperatures):
    # Ask the user for a temperature in Celcius
    celcius = float(input(f"Enter temperature {i+1} in Celcius: "))

# Convert Celcius to Fahrenheit using our function
    fahrenheit = celcius_to_fahrenheit(celcius)

# Add the converted temperature to the list
    temperatures.append(fahrenheit)
# Display the converted temperature
    print(f"{celcius} degrees Celcius is equal to {fahrenheit} degrees Fahrenheit.")

# Check whether the temperature is above or below the threshold
for temperature in temperatures:
    if temperature > 90:
        print(f"{temperature} degrees Fahrenheit is above the threshold.")
    elif temperature < 32:
        print(f"{temperature} degrees Fahrenheit is below the threshold.")
    else:
        print(f"{temperature} degrees Fahrenheit is within the threshold.")

    print () # Add a blank line between temperatures
```
