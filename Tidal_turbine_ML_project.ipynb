# 1. Project Problem and Intended AI/ML Outcome

## Project Problem

Tidal turbines convert the kinetic energy of moving seawater into useful electrical energy. However, turbine performance depends on operating conditions such as tidal flow velocity and tip-speed ratio (TSR). Determining the operating conditions that provide high turbine efficiency can require many simulations or experiments.

For this project, a simplified synthetic dataset is used to investigate the relationship between tidal turbine operating conditions and its power coefficient.

## Intended AI/ML Outcome

The aim is to develop a machine learning model that can predict the **power coefficient ($C_p$)** of a horizontal-axis tidal turbine from its operating conditions.

The initial input features will be:

- Tidal flow velocity ($V$)
- Tip-speed ratio (TSR)

The target variable will be:

- Power coefficient ($C_p$)

The trained machine learning model could eventually be used as a surrogate model to support faster tidal turbine performance prediction and design optimization.

> **Note:** The dataset used in this project is synthetic data generated using simplified tidal turbine equations. It is intended for demonstrating the AI/ML workflow and is not presented as experimental or CFD data.

# Turbine Specifications and Governing Equations

For this project, a hypothetical horizontal-axis tidal turbine is considered. The turbine dimensions are selected for demonstration purposes and do not represent a specific commercial turbine.

## Turbine Specifications

| Parameter | Value | Unit |

The annular swept area of the turbine is calculated as:

\[
A=\pi(R^2-R_h^2)
\]

where:

- \(A\) = effective swept area (m²)
- \(R\) = rotor radius (m)
- \(R_h\) = hub radius (m)

For the selected turbine:

\[
A=\pi(5^2-0.75^2)
\]

\[
A=76.77\;m^2
\]

## Governing Equations

### 1. Tip-Speed Ratio

The tip-speed ratio (TSR) is defined as:

\[
TSR=\frac{\omega R}{V}
\]

where:

- \(\omega\) = angular velocity (rad/s)
- \(R\) = rotor radius (m)
- \(V\) = tidal flow velocity (m/s)

The angular velocity can therefore be calculated as:

\[
\omega=\frac{TSR\times V}{R}
\]

### 2. Rotor Speed

Angular velocity is converted to revolutions per minute (RPM) using:

\[
RPM=\frac{\omega 60}{2\pi}
\]

### 3. Available Tidal Power

The kinetic power available in the tidal flow is:

\[
P_{available}=\frac{1}{2}\rho A V^3
\]

where:

- \(\rho\) = seawater density (kg/m³)
- \(A\) = effective swept area (m²)
- \(V\) = tidal flow velocity (m/s)

### 4. Turbine Power

The mechanical power extracted by the turbine is calculated using:

\[
P=C_pP_{available}
\]

or:

\[
P=C_p\left(\frac{1}{2}\rho A V^3\right)
\]

### 5. Torque

The turbine torque is calculated from:

\[
P=T\omega
\]

Therefore:

\[
T=\frac{P}{\omega}
\]

The synthetic dataset was generated by varying flow velocity and TSR and using a simplified relationship for \(C_p\). The resulting power, torque, rotor speed, and angular velocity were then calculated using the equations above.
